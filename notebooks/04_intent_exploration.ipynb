```python
import pandas as pd

amazon_df = pd.read_csv(
    "../data/processed/amazonhelp_cleaned.csv"
)

print("Shape:", amazon_df.shape)
print("\nColumns:")
print(amazon_df.columns.tolist())
print("\nFirst 5 rows:")
print(amazon_df.head())
```

```python
sample_conversation_ids = (
    amazon_df["conversation_id"]
    .dropna()
    .drop_duplicates()
    .sample(
        100,
        random_state=42
    )
)
```

```python
sample_conversations = amazon_df[
    amazon_df["conversation_id"].isin(
        sample_conversation_ids
    )
].copy()
```

```python
sample_conversations = sample_conversations.sort_values(
    ["conversation_id", "created_at"]
)
```

```python
for conversation_id, conversation in sample_conversations.groupby(
    "conversation_id"
):

    print("=" * 80)
    print(f"Conversation {conversation_id}")
    print("=" * 80)

    for _, row in conversation.iterrows():

        if row["inbound"] == True:
            role = "Customer"
        else:
            role = "AmazonHelp"

        print(f"\n{role}:")
        print(row["clean_text"])

    print("\n")
```

```python
sample_conversation_ids = (
    amazon_df["conversation_id"]
    .dropna()
    .drop_duplicates()
    .sample(
        500,
        random_state=42
    )
)

sample_conversations = amazon_df[
    amazon_df["conversation_id"].isin(
        sample_conversation_ids
    )
].copy()

sample_conversations = sample_conversations.sort_values(
    ["conversation_id", "created_at"]
)

for conversation_id, conversation in sample_conversations.groupby(
    "conversation_id"
):

    print("=" * 80)
    print(f"Conversation {conversation_id}")
    print("=" * 80)

    for _, row in conversation.iterrows():

        if row["inbound"] == True:
            role = "Customer"
        else:
            role = "AmazonHelp"

        print(f"\n{role}:")
        print(row["clean_text"])

    print("\n")
```

```python
# The 500 conversations you already inspected
inspected_500 = set(sample_conversation_ids)

# Get all available conversation IDs
all_conversation_ids = (
    amazon_df["conversation_id"]
    .dropna()
    .drop_duplicates()
)

# Remove the 500 already inspected
remaining_conversation_ids = all_conversation_ids[
    ~all_conversation_ids.isin(inspected_500)
]

# Randomly select 50 new conversations
new_50_conversation_ids = remaining_conversation_ids.sample(
    50,
    random_state=123
)

print("New conversations selected:", len(new_50_conversation_ids))

new_50_conversations = amazon_df[
    amazon_df["conversation_id"].isin(new_50_conversation_ids)
].copy()

new_50_conversations = new_50_conversations.sort_values(
    ["conversation_id", "created_at"]
)

new_50_conversations["turn_number"] = (
    new_50_conversations
    .groupby("conversation_id")
    .cumcount() + 1
)

for conversation_id, conversation in new_50_conversations.groupby(
    "conversation_id"
):

    print("=" * 80)
    print(f"Conversation {conversation_id}")
    print("=" * 80)

    for _, row in conversation.iterrows():

        role = "Customer" if row["inbound"] else "AmazonHelp"

        print(f"\n{role}:")
        print(row["clean_text"])

    print("\n")
```
