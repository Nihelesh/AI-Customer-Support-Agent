```python
import pandas as pd

spotify_df = pd.read_csv(
    "../data/processed/SpotifyCares.csv"
)

print("Shape:", spotify_df.shape)
print("\nColumns:")
print(spotify_df.columns.tolist())
print("\nFirst 5 rows:")
print(spotify_df.head())
```

```python
print("Total rows:", len(spotify_df))
print("Unique conversations:", spotify_df["conversation_id"].nunique())
```

```python
# Customer messages are identified by inbound == True
first_customer_messages = (
    spotify_df[
        spotify_df["inbound"] == True
    ]
    .groupby("conversation_id", as_index=False)
    .first()
)


# Keep only the columns needed for intent analysis
intent_df = first_customer_messages[
    [
        "conversation_id",
        "text"
    ]
].copy()


# Rename text column
intent_df = intent_df.rename(
    columns={
        "text": "customer_message"
    }
)


print("\nIntent discovery dataset:")
print(intent_df.head(20))

print("\nTotal conversations:")
print(len(intent_df))
```

```python
print(
    intent_df.sample(
        200,
        random_state=42
    )[
        ["conversation_id", "customer_message"]
    ].to_string(index=False)
)
```
